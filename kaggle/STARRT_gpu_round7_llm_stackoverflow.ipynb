# STARRT round 7: zero-shot LLMs on Stack Overflow (tool lists)

Fills the LLM gap: LLMs so far ran only on Djinni and Faruque. Dataset: **starrt-gpu6**. GPU T4 x2, Internet On.

Runs: Qwen2.5-3B zero-shot, then Phi-3.5-mini zero-shot, 5 folds, 40 per class. Expected ~6-8 h. Output: `starrt_gpu7_results.zip`.

```python
import glob, os, subprocess
hits = glob.glob('/kaggle/input/**/run_llm_zeroshot.py', recursive=True)
assert hits, 'Dataset not attached: Add Input -> starrt-gpu6'
CODE = os.path.dirname(hits[0]); print('code folder:', CODE)
import torch; print('GPUs:', torch.cuda.device_count(), [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])

def run(script, args=''):
    cmd = f'cd /kaggle/working && python -u {CODE}/{script} {args}'
    print('>>', cmd, flush=True); r = subprocess.run(cmd, shell=True)
    if r.returncode != 0: print('!! FAILED:', script, args)
```

```python
run('run_llm_zeroshot.py', '--datasets stackoverflow --folds 0,1,2,3,4 --per_class 40 --tag llm_qwen2.5-3b_zeroshot')
```

```python
run('run_llm_zeroshot.py', '--model microsoft/Phi-3.5-mini-instruct --end_text "<|end|>" --device_map auto --datasets stackoverflow --folds 0,1,2,3,4 --per_class 40 --tag llm_phi3.5-mini_zeroshot')
```
