```python
import glob, os, subprocess
hits = glob.glob('/kaggle/input/**/tune_models.py', recursive=True)
assert hits, 'Dataset not attached: Add Input -> starrt-gpu6'
CODE = os.path.dirname(hits[0]); print('code folder:', CODE)
import torch; print('GPUs:', torch.cuda.device_count(), [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())])

def run(script, args=''):
    cmd = f'cd /kaggle/working && python -u {CODE}/{script} {args}'
    print('>>', cmd, flush=True); r = subprocess.run(cmd, shell=True)
    if r.returncode != 0: print('!! FAILED:', script, args)
    return r.returncode
```

```python
# DistilBERT nested tuning on Stack Overflow (same protocol as the Djinni and Faruque runs)
rc = run('tune_models.py', '--family distilbert --datasets stackoverflow --model distilbert-base-uncased')
print('return code:', rc)
```
